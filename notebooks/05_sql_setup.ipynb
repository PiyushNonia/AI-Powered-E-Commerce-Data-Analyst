```python
import pandas as pd
import sqlite3
import os

# Load cleaned data
df = pd.read_csv(
    "../data/cleaned/cleaned_superstore.csv",
    encoding="latin1"
)

print("Cleaned data loaded!")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])

# Create database
db_path = "../data/superstore.db"

connection = sqlite3.connect(db_path)

# Create SQL table
df.to_sql(
    "sales",
    connection,
    if_exists="replace",
    index=False
)

connection.close()

print("SQLite database created successfully!")
print("Database location:", os.path.abspath(db_path))
```

```python
import sqlite3
import pandas as pd

connection = sqlite3.connect("../data/superstore.db")

print("Connected to SQLite database successfully!")
```

```python
tables = pd.read_sql_query(
    """
    SELECT name
    FROM sqlite_master
    WHERE type='table';
    """,
    connection
)
display(tables)
```

```python
query = """
SELECT *
FROM sales
LIMIT 10;
"""
result = pd.read_sql_query(query, connection)
display(result)
```

```python
query = """
SELECT
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY Category
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    Region,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY Region
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Product Name"
ORDER BY Total_Sales DESC
LIMIT 10;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    "Product Name",
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY "Product Name"
HAVING Total_Profit < 0
ORDER BY Total_Profit ASC
LIMIT 10;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    Segment,
    COUNT(DISTINCT "Customer ID") AS Customers,
    COUNT(DISTINCT "Order ID") AS Orders,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit
FROM sales
GROUP BY Segment
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    Discount,
    ROUND(SUM(Sales), 2) AS Total_Sales,
    ROUND(SUM(Profit), 2) AS Total_Profit,
    COUNT(*) AS Transactions
FROM sales
GROUP BY Discount
ORDER BY Discount;
"""

result = pd.read_sql_query(query, connection)

display(result)
```

```python
query = """
SELECT
    Category,
    ROUND(SUM(Sales), 2) AS Sales,
    ROUND(SUM(Profit), 2) AS Profit,
    ROUND(
        SUM(Profit) * 100.0 /
        NULLIF(SUM(Sales), 0),
        2
    ) AS Profit_Margin_Percent
FROM sales
GROUP BY Category
ORDER BY Profit_Margin_Percent DESC;
"""

result = pd.read_sql_query(query, connection)

display(result)
```
