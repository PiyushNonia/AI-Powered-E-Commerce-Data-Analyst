```python
# ============================================================
# AI E-COMMERCE DATA ANALYSIS
# 01 - DATA UNDERSTANDING
# ============================================================

# 1. Import Libraries
import pandas as pd
import numpy as np


# 2. Load Dataset
df = pd.read_csv(
    "../data/raw/Sample - Superstore.csv",
    encoding="latin1"
)

print("Dataset loaded successfully!")


# 3. Dataset Size
print("Number of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])


# 4. View First 10 Rows
print("\nFirst 10 Rows:")
display(df.head(10))


# 5. Column Names
print("\nColumns in Dataset:")
for column in df.columns:
    print(column)


# 6. Dataset Information
print("\nDataset Information:")
df.info()


# 7. Missing Values
print("\nMissing Values in Each Column:")
missing_values = df.isnull().sum()
print(missing_values)


# 8. Duplicate Records
print("\nDuplicate Rows:", df.duplicated().sum())


# 9. Statistical Summary
print("\nStatistical Summary:")
display(df.describe().T)


# 10. Unique Values in Important Columns
print("\nCategories:")
print(df["Category"].unique())

print("\nRegions:")
print(df["Region"].unique())

print("\nSegments:")
print(df["Segment"].unique())


# 11. Number of Unique Customers
print("\nUnique Customers:", df["Customer ID"].nunique())


# 12. Number of Unique Orders
print("Unique Orders:", df["Order ID"].nunique())


# 13. Number of Unique Products
print("Unique Products:", df["Product Name"].nunique())


# 14. Final Dataset Overview
print("\n========== DATASET OVERVIEW ==========")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
print("Unique Orders:", df["Order ID"].nunique())
print("Unique Customers:", df["Customer ID"].nunique())
print("Unique Products:", df["Product Name"].nunique())
print("Duplicate Rows:", df.duplicated().sum())
print("Total Missing Values:", df.isnull().sum().sum())
```
